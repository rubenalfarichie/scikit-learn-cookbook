## Ringkasan Bab & Penjelasan Teoretis
Bab ini membahas filosofi desain dasar dari pustaka `scikit-learn` yang berpusat pada empat prinsip utama: **konsistensi, kesederhanaan, modularitas, dan *reusability***. Seluruh komponen di dalam `scikit-learn` menggunakan antarmuka (API) yang seragam sehingga memudahkan transisi antar algoritma machine learning.

### 1. Estimators (`fit` dan `predict`)
**Estimator** adalah objek berbasis *Object-Oriented Programming* (OOP) yang mengimplementasikan algoritma pembelajaran dari data:
* `fit(X, y)`: Melatih model dengan mempelajari parameter optimal dari matriks fitur $X$ dan vektor target $y$ (pada *supervised learning*) atau hanya dari $X$ (pada *unsupervised learning*).
* `predict(X_new)`: Menghasilkan prediksi pada data baru berdasarkan parameter yang telah dipelajari selama tahap `fit()`.
* `fit_predict(X)`: Jalan pintas API yang menggabungkan pelatihan dan prediksi sekaligus pada dataset yang sama, umumnya diterapkan pada skenario *unsupervised learning* seperti klasterisasi (`KMeans`).

```python
from sklearn.linear_model import LinearRegression
import numpy as np

# Contoh data: Matriks fitur X dan nilai target y
X = np.array([[1], [2], [3], [4], [5]])
y = np.array([1, 2, 3, 3.5, 5])

# Inisialisasi dan latih model (Estimator)
model = LinearRegression()
model.fit(X, y)

# Prediksi nilai untuk data baru
X_new = np.array([[6], [7]])
predictions = model.predict(X_new)
print("Predictions:", predictions)
```

```python
from sklearn.cluster import KMeans

# Contoh penggunaan fit_predict pada Unsupervised Learning (KMeans)
X = np.array([[1], [2], [3], [4], [5]])
kmeans = KMeans(n_clusters=2, random_state=42)
labels = kmeans.fit_predict(X)
print("Cluster Labels:", labels)
```

### 2. Transformers (`transform` dan `fit_transform`)
**Transformer** adalah komponen yang memodifikasi data melalui penyekalaan, normalisasi, atau *encoding* sebelum tahap pemodelan:
* `fit(X)` mempelajari parameter statistik dari data (seperti rata-rata $\mu$ dan simpangan baku $\sigma$ pada `StandardScaler`).
* `transform(X)` menerapkan perubahan skala menggunakan parameter tersebut (menghasilkan *z-score*).
* `fit_transform(X)` menggabungkan kedua langkah untuk efisiensi pada **data latih**. Pada **data uji (*test set*)**, kita hanya boleh memanggil `transform()` agar tetap menggunakan parameter statistik dari data latih dan mencegah kebocoran data (*data leakage*).

```python
from sklearn.preprocessing import StandardScaler
import numpy as np

X = np.array([[1, 2], [3, 4], [5, 6]])

# Menggunakan fit() dan transform() secara terpisah
scaler = StandardScaler()
scaler.fit(X)
X_scaled = scaler.transform(X)
print("Scaled Data (fit + transform):\n", X_scaled)

# Menggunakan shortcut fit_transform()
scaler_shortcut = StandardScaler()
X_scaled_shortcut = scaler_shortcut.fit_transform(X)
print("\nScaled Data (fit_transform):\n", X_scaled_shortcut)
```

### 3. Atribut Umum, Evaluasi (`score`), dan Penyetelan Hyperparameter
* **Atribut Terlatih**: Atribut dengan akhiran garis bawah (*trailing underscore*) seperti `coef_` (kemiringan garis) dan `intercept_` (titik potong sumbu-y) menyimpan parameter hasil pelatihan model.
* **Metode `score()`**: Mengevaluasi performa model secara cepat menggunakan metrik bawaan ($R^2$ untuk regresi, akurasi untuk klasifikasi).
* **Manajemen Hyperparameter & Metadata Routing**: Metode `set_params()` dan `get_params()` memungkinkan pengaturan hyperparameter secara terprogram sebelum dioptimalkan dengan `GridSearchCV()`. Selain itu, fitur *metadata routing* mengatur pengiriman metadata (seperti *sample weights*) antara objek *router* dan *consumer* di dalam *pipeline*.

```python
from sklearn.linear_model import LinearRegression
from sklearn.ensemble import RandomForestClassifier
import numpy as np

X = np.array([[1], [2], [3], [4], [5]])
y = np.array([1, 2, 3, 3.5, 5])

model = LinearRegression()
model.fit(X, y)

print("Coefficients:", model.coef_)
print("Intercept:", model.intercept_)
print("Model R-squared:", model.score(X, y))

# Mengatur dan memeriksa hyperparameter dengan set_params() dan get_params()
rf_model = RandomForestClassifier()
rf_model.set_params(n_estimators=100, max_depth=10, random_state=42)
print("\nUpdated RandomForest Parameters:\n", rf_model.get_params())
```
